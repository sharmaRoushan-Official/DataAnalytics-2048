Indexing and slicing of matrix

```python
import numpy as np
arr = np.array([10,20,30])
print(arr)
```

```python
#### positive indexing of matrix
print(arr[0])
print(arr[1])
print(arr[2])
```

```python
#### Negative Indexing of Matrix
print(arr[-1])
print(arr[-2])
print(arr[-3])
```

### Slicing of list

syntax:

arr[start:stop:stride]

```python
arr = np.array([10,20,30,40,50])
arr
```

```python
print(arr[2:])
```

```python
print(arr[4:])
```

```python
print(arr[2:4])
```

```python
print(arr[:3:])
```

```python
print(arr[2:5:])
```

#### Boolean Indexing

Boolean indexin selects element that satisfy a condition




```python
marks = np.array([45,78,92,33,65,88])

print(marks)
```

```python
print(marks.shape)
```

```python
marks1 = marks[marks>50]
print(marks1)
```

```python
marks2 = marks[marks>80]
marks2
```

```python
# >50 and <80
marks3 = marks[(marks>50) & (marks<80)]
print(marks3)
```

```python
marks4 = marks[(marks>80)&(marks<90)]
marks4
```
