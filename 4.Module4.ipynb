### Population Variance


````python
import numpy as np 
from scipy import stats

marks = np.array([47,65,75,80,85,85,85,85])
print(marks)

print("Sum:",sum(marks))
print("Count:",len(marks))
print("Mean:",np.mean(marks))
print("Median:",np.median(marks))
print("Mode:",stats.mode(marks))
print("Maximum:",max(marks))
print("Minimum:",min(marks))
print("Standard Deviation:",np.std(marks))
print("Variance:",np.var(marks))
````

````python
data = np.array([
    [[10, 20],
    [30, 40]],

    [[50,60],
    [70,80]]
])
print(data)
print("Varience:",np.var(data))
print("Standard Deviation:",np.std(data)) 
print("Cross Wise Standard Deviation:",np.std(data,axis=0))
print("Column wise Standard Deviation:",np.std(data,axis=1))
print("Rows Standard Deviation:",np.std(data,axis=2))
````
