### Population Variance


```python
import numpy as np 
from scipy import stats

marks = np.array([47,65,75,80,85,85,85,85])
print(marks)

print("Sum:",sum(marks))
print("Count:",len(marks))
print("Mean:",np.mean(marks))
print("Median:",np.median(marks))
print("Mode:",stats.mode(marks))
print("Maximum:",max(marks))
print("Minimum:",min(marks))
print("Standard Deviation:",np.std(marks))
print("Variance:",np.var(marks))
```

```python
data = np.array([2,4,6])
print(data)

print("Population variance:",np.var(data))
```

```python
data1 = [2,8,10,12,14,15]
matData = np.array(data1)
print(matData)
print("Variance:",np.var(matData))
```
