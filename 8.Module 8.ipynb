### 8. Module 8 - Searching and Sorting Array

28. Where

Selects values according to conditions:

````python
import numpy as np

marks = np.array([34,78,45,90,28])
print(marks)

result = np.where(marks>=40,"Pass","Fail")
print(result)
````

````python
# 2-D Data 

matrix = np.array([
    [10,11,12,13],
    [14,15,16,17]
])
print(matrix)
````

````python
result = np.where(matrix%2==0,"Even","Odd")
result
````

### It can also return the indices where a condition is true

````python
marks = np.array([34,78,45,90,28])
print(marks)

indexValue = np.where(marks>40)
indexValue
````

````python
matrix = np.array([
    [10,11,12,13],
    [14,15,16,17]
])
print(matrix)

result = np.where(matrix%2==0)
print(result)
````

## Sort()  and argsort()

np.sort() return sorted values



````python
arr = np.array([10,12,15,14,17,18,12])
print(arr)
````

````python
print("Sorted Array:",np.sort(arr))
````

````python
arr = np.array([10,12,15,14,17,18,12])
print(arr)
print("Argsort:",np.argsort(arr)) # position value.
````

### max() and min(), argmax() and argmin()

argmin and arg max() Return the index of minimum and maximum value

````python
arr = np.array([10,12,15,14,17,18,12])
print(arr)

print("Minimum:",min(arr))
print("Index Position:",np.argmin(arr))

print("Maximum Value:",max(arr))
print("Index Position:",np.argmax(arr))
````
