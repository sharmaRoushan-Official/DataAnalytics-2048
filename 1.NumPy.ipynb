### Types of Arrays in Numpy

In NumPy, arrays are used to store multiple values in a single variable. NumPy array are called ndarrays (N-dimensional arrays)

The main types of NumPy arrays based on dimensions are:

1. 0-D Array - [Scalar]
2. 1-D Array - [One -dimensional]
3. 2-D Array - [Two-dimensional]
4. 3-D Array - [Three-dimensional]
5. N-D Array - [Multidimensional]

### 1. O-D Array [ Scalar ]

A O-D Dimension contains only a single value

````python
arr = np.array(10)
print(arr)
print(arr.ndim) # dimension
print(arr.shape) # 0-D 
````

Notes:

* ndim returns the number of dimenstions.
* shape returns the size of each dimension.

### 2. Data Types

# 2. Common NumPy Data Types

NumPy supports a wide range of data types, giving you precise control over memory and bits.

| Category | NumPy Type Examples | Character Code | Description |
|---|---|---|---|
| Integers | `int8`, `int16`, `int32`, `int64` | `'i'` | Signed whole numbers |
| Unsigned Integers | `uint8`, `uint16`, `uint32`, `uint64` | `'u'` | Positive-only integers (including zero) |
| Floats | `float16`, `float32`, `float64` | `'f'` | Decimal numbers |
| Booleans | `bool_` | `'b'` | `True` or `False` |
| Strings | `str_`, `string_` | `'S'`, `'U'` | Text data (fixed-length Unicode) |
| Objects | `object_` | `'O'` | Fallback for mixed Python objects |

## Examples

```python
import numpy as np

# Integers
a = np.array([10, 20, 30], dtype=np.int32)

# Unsigned Integers
b = np.array([10, 20, 30], dtype=np.uint8)

# Floats
c = np.array([10.5, 20.5, 30.5], dtype=np.float64)

# Booleans
d = np.array([True, False, True], dtype=np.bool_)

# Strings
e = np.array(["Python", "NumPy"], dtype=str)

# Objects
f = np.array([10, "Python", 20.5], dtype=object)

print(a.dtype)
print(b.dtype)
print(c.dtype)
print(d.dtype)
print(e.dtype)
print(f.dtype)
```

## Important Points

- `int` → Whole numbers.
- `uint` → Non-negative whole numbers.
- `float` → Decimal numbers.
- `bool` → True or False.
- `str` → Text values.
- `object` → Stores Python objects of different types.

**Note:** The `dtype` parameter specifies the data type of NumPy array elements.

#### NumPy Zeros Array

1. What is a Zeros Array?

In NumPy, `np.zeros()` is used to create an array in which all elemnets are intialized to Zero (0)

It is useful when we need an empty-initialized numerical array before storing and processing data.



````python
arr = np.zeros(3)   # int64
print(arr)
print(arr.dtype)
````

### 1. What is a Ones Array?

In NumPy, np.ones() is used to create array in which all elements are initialized to 1.
