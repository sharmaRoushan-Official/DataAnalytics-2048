NumPy Prform element wise arithematic operations when the arrays have compatible shape.

```python
import numpy as np
a = np.array([10,20,30])
b = np.array([2,4,5])

print(a)
print(b)
```

```python
# Addition 
a = np.array([10,20,30])
b = np.array([2,4,5])

print(a)
print(b)
print(a+b)  # by default - row wise
```

```python
a = np.array([10,20,30])
b = np.array([2,4,5])

print(a)
print(b)
print(a-b)
```

```python
a = np.array([10,20,30])
b = np.array([2,4,5])

print(a)
print(b)
print(a*b)
```

####  Comparison and Logical Operations


```python
arr = np.array([10,25,40,55,70])
print(arr)
print(arr>30)
```

```python
print(arr<50)
```

```python
print(arr==40)
```

Notes:

Comparison function produces boolean arrays. 
These boolean arrays can be used to filter, count or replaces values.

### 14. Universal Function (ufuncs)


```python
arr = np.array([1,4,9,16,25])
print(arr)
```

```python
print(np.sqrt(arr))
```

```python
print(np.square(arr))
```

```python
print(np.exp(arr))
```

```python
print(np.log(arr))
```

```python
print(arr)
print(np.cbrt(arr))
```

|Category|Function|
|----|----|
